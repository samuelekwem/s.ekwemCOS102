```python
import tkinter as tk
from tkinter import messagebox
import csv
import os

# Create jobs.csv if it doesn't exist
if not os.path.exists("jobs.csv"):
    with open("jobs.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Job Title", "Company", "Location", "Salary"])


# ---------------- POST JOB FUNCTION ----------------
def post_job():
    title = title_entry.get()
    company = company_entry.get()
    location = location_entry.get()
    salary = salary_entry.get()

    if title == "" or company == "" or location == "" or salary == "":
        messagebox.showerror("Error", "All fields are required")
        return

    with open("jobs.csv", "a", newline="") as file:
        writer = csv.writer(file)
        writer.writerow([title, company, location, salary])

    messagebox.showinfo("Success", "Job Posted Successfully")

    title_entry.delete(0, tk.END)
    company_entry.delete(0, tk.END)
    location_entry.delete(0, tk.END)
    salary_entry.delete(0, tk.END)


# ---------------- VIEW JOBS FUNCTION ----------------
def view_jobs():
    jobs_window = tk.Toplevel(root)
    jobs_window.title("Posted Jobs")
    jobs_window.geometry("600x400")

    text_area = tk.Text(jobs_window, width=70, height=20)
    text_area.pack(pady=20)

    with open("jobs.csv", "r") as file:
        reader = csv.reader(file)

        for row in reader:
            text_area.insert(tk.END, f"""
Job Title : {row[0]}
Company   : {row[1]}
Location  : {row[2]}
Salary    : {row[3]}
-----------------------------------------
""")


# ---------------- MAIN WINDOW ----------------
root = tk.Tk()
root.title("Employer Dashboard")
root.geometry("500x450")
root.configure(bg="lightblue")

title_label = tk.Label(root, text="Employer Dashboard",
                       font=("Arial", 20, "bold"),
                       bg="lightblue")
title_label.pack(pady=10)

# Job Title
tk.Label(root, text="Job Title", bg="lightblue",
         font=("Arial", 12)).pack()
title_entry = tk.Entry(root, width=40)
title_entry.pack(pady=5)

# Company
tk.Label(root, text="Company Name", bg="lightblue",
         font=("Arial", 12)).pack()
company_entry = tk.Entry(root, width=40)
company_entry.pack(pady=5)

# Location
tk.Label(root, text="Location", bg="lightblue",
         font=("Arial", 12)).pack()
location_entry = tk.Entry(root, width=40)
location_entry.pack(pady=5)

# Salary
tk.Label(root, text="Salary", bg="lightblue",
         font=("Arial", 12)).pack()
salary_entry = tk.Entry(root, width=40)
salary_entry.pack(pady=5)

# Buttons
post_button = tk.Button(root,
                        text="Post Job",
                        command=post_job,
                        bg="green",
                        fg="white",
                        width=20)

post_button.pack(pady=10)

view_button = tk.Button(root,
                        text="View Posted Jobs",
                        command=view_jobs,
                        bg="blue",
                        fg="white",
                        width=20)

view_button.pack(pady=10)

root.mainloop()
```
