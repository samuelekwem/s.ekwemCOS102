```python
from abc import ABC, abstractmethod

class Product(ABC):

    def item_list(self, rate):
        print("Amount submitted:", rate)

    @abstractmethod
    def product(self, rate):
        pass


class Goods(Product):

    def product(self, rate):
        print("Product rate:", rate)


r = Goods()
r.product(3)
r.item_list(2)
```

```python
from abc import ABC, abstractmethod

class Geometric(ABC):

    @abstractmethod
    def volume(self):
        pass


class Rect(Geometric):
    length = 4
    width = 6
    height = 6

    def volume(self):
        return self.length * self.width * self.height


class Sphere(Geometric):
    radius = 8

    def volume(self):
        return (4/3) * 3.142 * self.radius**3


class Cube(Geometric):
    edge = 5

    def volume(self):
        return self.edge**3


rr = Rect()
ss = Sphere()
cc = Cube()

print("Volume of Rectangle:", rr.volume())
print("Volume of Sphere:", ss.volume())
print("Volume of Cube:", cc.volume())
```

```python
from abc import ABC, abstractmethod

class Invoice(ABC):

    def final_bill(self, pay):
        print("Purchase of product:", pay)

    @abstractmethod
    def invoice(self, pay):
        pass


class Paycheque(Invoice):

    def invoice(self, pay):
        print("Paycheque of:", pay)


class CardPayment(Invoice):

    def invoice(self, pay):
        print("Pay through card of:", pay)


aa = Paycheque()
aa.invoice(6500)
aa.final_bill(6500)

bb = CardPayment()
bb.invoice(2600)
bb.final_bill(2600)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def move(self):
        pass


class Human(Animal):

    def move(self):
        print("I can walk and run")


class Snake(Animal):

    def move(self):
        print("I can crawl")


class Dog(Animal):

    def move(self):
        print("I can walk and run")


class Lion(Animal):

    def move(self):
        print("I can run")


human = Human()
human.move()

snake = Snake()
snake.move()

dog = Dog()
dog.move()

lion = Lion()
lion.move()
```

```python
import abc
from abc import ABC

class Zinc(ABC):

    def rk(self):
        print("Class Zinc is an Abstract Base Class")


class Tin(Zinc):

    def rk(self):
        super().rk()
        print("Class Tin is a subclass")


r = Tin()
r.rk()
```

```python
from abc import ABC, abstractmethod

class Bank(ABC):

    def branch(self, naira):
        print("Fees submitted:", naira)

    @abstractmethod
    def bank(self, naira):
        pass


class Private(Bank):

    def bank(self, naira):
        print("Total Naira Value here:", naira)


class Public(Bank):

    def bank(self, naira):
        print("Total Naira Value here:", naira)


p1 = Private()
p1.bank(5000)

p2 = Public()
p2.bank(2000)

p2.branch(3500)
```
