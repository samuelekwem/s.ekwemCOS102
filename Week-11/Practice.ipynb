```python
class Employee:

    # constructor
    def __init__(self, name, salary, project):
        self.name = name
        self.salary = salary
        self.project = project

    # method to display employee details
    def show(self):
        print("\nName:", self.name)
        print("Salary: N" + str(self.salary))

    # method
    def work(self):
        print(self.name, "is working on", self.project)


name = input("Enter your name: ")
salary = int(input("How much do you earn: "))
project = input("What project are you working on: ")

emp = Employee(name, salary, project)

emp.show()
emp.work()
```

```python
# Base class
class Company:

    def __init__(self):
        self._project = "Blockchain Development"


# Child class
class Employee(Company):

    def __init__(self, name):
        self.name = name
        Company.__init__(self)

    def show(self):
        print("Employee name:", self.name)
        print("Working on project:", self._project)


c = Employee("Prince Ibekwe")

c.show()

print("Project:", c._project)
```

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.__age = age

    # getter
    def get_age(self):
        return self.__age

    # setter
    def set_age(self, age):
        self.__age = age


stud = Student("Michael Illoba", 34)

print("Name:", stud.name)
print("Age:", stud.get_age())

stud.set_age(26)

print("\nUpdated Age:", stud.get_age())
```

```python
class Student:

    def __init__(self, name, roll_no, age):
        self.name = name
        self.__roll_no = roll_no
        self.__age = age

    def show(self):
        print("Student Details:", self.name, self.__roll_no)

    def get_roll_no(self):
        return self.__roll_no

    def set_roll_no(self, number):
        if number > 50:
            print("Invalid roll no. Please set correct roll number")
        else:
            self.__roll_no = number


info = Student("David Usim", 10, 15)

info.show()

info.set_roll_no(120)

info.set_roll_no(25)

info.show()
```
