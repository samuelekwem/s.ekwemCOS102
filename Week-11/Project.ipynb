```python
from abc import ABC, abstractmethod

class coup_de_ecriva(ABC):

    @abstractmethod
    def Fan_Page(self):
        pass


class FC_Cirok(coup_de_ecriva):

    def Fan_Page(self):
        print("Welcome to FC_Cirok")


class Madiba_FC(coup_de_ecriva):

    def Fan_Page(self):
        print("Welcome to Madiba_FC")


class Blue_Jay_FC(coup_de_ecriva):

    def Fan_Page(self):
        print("Welcome to Blue_Jay_FC")


class TSG_Walker(coup_de_ecriva):

    def Fan_Page(self):
        print("Welcome to TSG_Walker")


club = input("Enter club name: ")

if club.lower() == "fc_cirok":
    obj = FC_Cirok()

elif club.lower() == "madiba_fc":
    obj = Madiba_FC()

elif club.lower() == "blue_jay_fc":
    obj = Blue_Jay_FC()

elif club.lower() == "tsg_walker":
    obj = TSG_Walker()

else:
    print("Club not found")
    obj = None

if obj:
    obj.Fan_Page()
```

```python
from abc import ABC, abstractmethod


class Vendor(ABC):

    @abstractmethod
    def menu(self):
        pass


class CooperativeHostel(Vendor):

    def menu(self):
        print("\nCOOPERATIVE CAFETERIA")
        print("Jollof Rice and Stew - N200")
        print("White Rice and Stew - N200")
        print("Fried Rice - N200")
        print("Salad - N100")
        print("Plantain - N100")


class FaithHostel(Vendor):

    def menu(self):
        print("\nFAITH HOSTEL CAFETERIA")
        print("Fried Rice - N400")
        print("White Rice and Stew - N400")
        print("Jollof Rice - N400")
        print("Beans - N200")
        print("Chicken - N1000")


class StudentCenter(Vendor):

    def menu(self):
        print("\nSTUDENT CENTER CAFETERIA")
        print("Chicken Fried Rice - N800")
        print("Pomo Sauce - N300")
        print("Spaghetti Jollof - N500")
        print("Amala/Ewedu - N500")
        print("Semo with Soup - N500")


vendor = input(
    "Choose Vendor (Faith Hostel / Cooperative Hostel / Student Center): "
)

if vendor.lower() == "faith hostel":
    obj = FaithHostel()

elif vendor.lower() == "cooperative hostel":
    obj = CooperativeHostel()

elif vendor.lower() == "student center":
    obj = StudentCenter()

else:
    print("Vendor not found")
    obj = None

if obj:
    obj.menu()
```
