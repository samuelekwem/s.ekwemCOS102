```python
class Person(object):

    # Constructor
    def __init__(self, name):
        self.name = name

    # To get name method
    def getname(self):
        return self.name

    # To check if this person is an employee
    def isEmployee(self):
        return False


# Inherited or Subclass
class Employee(Person):

    # Here employee method returns true
    def isEmployee(self):
        return True


# Object Instantiation
emp = Person("Segun Da-Silver")
print(emp.getname(), emp.isEmployee())

emp = Employee("Omotayo Ayeni")
print(emp.getname(), emp.isEmployee())
```

```python
# Parent class
class Person(object):

    # Constructor
    def __init__(self, name, idnumber):
        self.name = name
        self.idnumber = idnumber

    def display(self):
        print("Name:", self.name)
        print("ID No.:", self.idnumber)


# Child class
class Employee(Person):

    def __init__(self, name, idnumber, salary, post):
        self.salary = salary
        self.post = post

        # Invoking parent constructor
        Person.__init__(self, name, idnumber)

    def display(self):
        Person.display(self)
        print("Salary:", self.salary)
        print("Post:", self.post)


# Object creation
a = Employee('Adedoyin Adeniji', 886012, 200000, "Intern")

# Calling function
a.display()
```

```python
# Base class
class Parent:
    def func1(self):
        print("This function is in parent class.")


# Derived class
class Child(Parent):
    def func2(self):
        print("This function is in child class.")


# Object Instance
obj = Child()

obj.func1()
obj.func2()
```

```python
# Base class 1
class Mother:
    mothername = "Caroline Aina"

    def mother(self):
        print(self.mothername)


# Base class 2
class Father:
    fathername = "Paul Abiodun"

    def father(self):
        print(self.fathername)


# Derived class
class Son(Mother, Father):

    def parents(self):
        print("Father :", self.fathername)
        print("Mother :", self.mothername)


# Object Instance
s1 = Son()
s1.parents()
```

```python
# Base class
class Grandfather:

    def __init__(self, grandfathername):
        self.grandfathername = grandfathername


# Intermediate class
class Father(Grandfather):

    def __init__(self, fathername, grandfathername):
        self.fathername = fathername

        # Invoking constructor of Grandfather
        Grandfather.__init__(self, grandfathername)


# Derived class
class Son(Father):

    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname

        # Invoking constructor of Father
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print("Grandfather name :", self.grandfathername)
        print("Father name :", self.fathername)
        print("Son name :", self.sonname)


# Object Instance
s1 = Son('Gbenga', 'Oluwafemi', 'Adeoke')

print(s1.grandfathername)
s1.print_name()
```

```python
# Base class
class Parent:
    def func1(self):
        print("This function is in parent class.")


# Derived class 1
class Child1(Parent):
    def func2(self):
        print("This function is in child 1.")


# Derived class 2
class Child2(Parent):
    def func3(self):
        print("This function is in child 2.")


# Object Instantiation
object1 = Child1()
object2 = Child2()

object1.func1()
object1.func2()

object2.func1()
object2.func3()
```

```python
# Python program to demonstrate hybrid inheritance

class School():
    def func1(self):
        print("This function is in school.")


class Student1(School):
    def func2(self):
        print("This function is in student 1.")


class Student2(School):
    def func3(self):
        print("This function is in student 2.")


class Student3(Student1, School):
    def func4(self):
        print("This function is in student 3.")


# Object Instantiation
obj = Student3()

obj.func1()
obj.func2()
obj.func4()
```

```python
# A simple Python function to demonstrate polymorphism

def add(x, y, z=0, w=0):
    return x + y + z + w


# Driver code
print(add(2, 3))
print(add(2, 3, 4))
print(add(2, 3, 5, 6))
```

```python
class Nigeria():

    def capital(self):
        print("Lagos is the capital of Nigeria.")

    def language(self):
        print("English is the official language of Nigeria.")

    def type(self):
        print("Nigeria is the giant of Africa.")


class Togo():

    def capital(self):
        print("Lome is the capital of Togo.")

    def language(self):
        print("French is the primary language of Togo.")

    def type(self):
        print("Togo is known for its palm-lined beaches and hilltop villages.")


obj_naija = Nigeria()
obj_togo = Togo()

for country in (obj_naija, obj_togo):
    country.capital()
    country.language()
    country.type()
```

```python
class Bird:

    def intro(self):
        print("There are many types of birds.")

    def flight(self):
        print("Most birds can fly but some cannot.")


class Sparrow(Bird):

    def flight(self):
        print("Sparrows can fly.")


class Ostrich(Bird):

    def flight(self):
        print("Ostriches cannot fly.")


obj_bird = Bird()
obj_spr = Sparrow()
obj_ost = Ostrich()

obj_bird.intro()
obj_bird.flight()

obj_spr.intro()
obj_spr.flight()

obj_ost.intro()
obj_ost.flight()
```

```python
from tkinter import *


# Parent Class
class Zenith:

    def mutual_services(self):
        return [
            "Lines of Credit",
            "Investment Management and Accounts",
            "Insurance"
        ]


# Retail Banking Class
class RetailBanking(Zenith):

    def unique_services(self):
        return [
            "Retirement and Education Accounts",
            "Loans and Mortgages",
            "Checking and Saving"
        ]


# Global Banking Class
class GlobalBanking(Zenith):

    # Polymorphism (method overriding)
    def mutual_services(self):
        return [
            "Global Financial Advisory",
            "International Transactions"
        ]

    def unique_services(self):
        return [
            "Multi-currency Management Services and Products",
            "Foreign Currency Accounts",
            "Foreign Currency Credit Cards",
            "Transborder Advisory Services",
            "Liquidity Management"
        ]


# Commercial Banking Class
class CommercialBanking(Zenith):

    def unique_services(self):
        return [
            "Advisory Services"
        ]


# Function to display services
def display_services():

    employee = employee_entry.get()
    division = division_entry.get().lower()

    output.delete("1.0", END)

    if division == "retail":
        obj = RetailBanking()

    elif division == "global":
        obj = GlobalBanking()

    elif division == "commercial":
        obj = CommercialBanking()

    else:
        output.insert(END, "Invalid Division Entered!")
        return

    output.insert(END, f"Employee Name: {employee}\n")
    output.insert(END, f"Division: {division.capitalize()} Banking\n\n")

    output.insert(END, "Mutual Services:\n")
    for service in obj.mutual_services():
        output.insert(END, f"• {service}\n")

    output.insert(END, "\nUnique Services:\n")
    for service in obj.unique_services():
        output.insert(END, f"• {service}\n")


# GUI Window
root = Tk()
root.title("Zenith Bank Services")
root.geometry("600x500")
root.config(bg="lightblue")


# Heading
title = Label(
    root,
    text="Zenith Bank Nigeria",
    font=("Arial", 18, "bold"),
    bg="lightblue"
)
title.pack(pady=10)


# Employee Name
employee_label = Label(
    root,
    text="Enter Employee Name:",
    font=("Arial", 12),
    bg="lightblue"
)
employee_label.pack()

employee_entry = Entry(root, width=40)
employee_entry.pack(pady=5)


# Division
division_label = Label(
    root,
    text="Enter Division (Retail, Global, Commercial):",
    font=("Arial", 12),
    bg="lightblue"
)
division_label.pack()

division_entry = Entry(root, width=40)
division_entry.pack(pady=5)


# Button
submit_btn = Button(
    root,
    text="Display Services",
    command=display_services,
    bg="navy",
    fg="white",
    font=("Arial", 12, "bold")
)
submit_btn.pack(pady=15)


# Output Box
output = Text(root, width=70, height=15)
output.pack(pady=10)


# Run Program
root.mainloop()
```
