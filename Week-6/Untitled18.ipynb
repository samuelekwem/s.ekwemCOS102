```python
from tkinter import *
import random


class Employee:

    employees = [
        "Mary Evans", "Eyo Ishan", "Durojaiye Dare",
        "Adams Ali", "Andrew Ugwu", "Stella Mankinde",
        "Jane Akibo", "Ago James", "Michell Taiwo",
        "Abraham Jones", "Nicole Anide", "Kosi Korso",
        "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"
    ]

    tasks = [
        "Loading",
        "Transporting",
        "Reviewing Orders",
        "Customer Service",
        "Delivering Items"
    ]

    def __init__(self, name):
        self.name = name

    def check_employee(self):
        return self.name in Employee.employees

    def take_attendance(self):
        return f"Attendance taken for {self.name}."

    def assign_task(self):
        task = random.choice(Employee.tasks)
        return f"Assigned Task: {task}"

    def refuse_access(self):
        return "Access Denied! You are not an employee."


# Function
def verify_employee():

    name = entry.get()

    emp = Employee(name)

    output.delete("1.0", END)

    if emp.check_employee():
        output.insert(END, f"Welcome {name}!\n")
        output.insert(END, emp.take_attendance() + "\n")
        output.insert(END, emp.assign_task())

    else:
        output.insert(END, emp.refuse_access())


# GUI Window
root = Tk()
root.title("Employee Verification System")
root.geometry("500x400")
root.config(bg="lightblue")


# Heading
heading = Label(
    root,
    text="Employee Verification System",
    font=("Arial", 16, "bold"),
    bg="lightblue"
)
heading.pack(pady=10)


# Name Label
label = Label(
    root,
    text="Enter Employee Name:",
    font=("Arial", 12),
    bg="lightblue"
)
label.pack()


# Entry Box
entry = Entry(root, width=40)
entry.pack(pady=10)


# Button
button = Button(
    root,
    text="Verify",
    command=verify_employee,
    bg="navy",
    fg="white"
)
button.pack(pady=10)


# Output Box
output = Text(root, width=55, height=10)
output.pack(pady=10)


# Run App
root.mainloop()
```

```python
from tkinter import *


class DeliveryService:

    def __init__(self, location, weight):
        self.location = location
        self.weight = weight

    def calculate_price(self):

        if self.location == "PAU":

            if self.weight >= 10:
                return 2000
            else:
                return 1500

        elif self.location == "Epe":

            if self.weight >= 10:
                return 5000
            else:
                return 4000

        else:
            return None


# Function
def check_price():

    location = location_entry.get()
    weight = float(weight_entry.get())

    delivery = DeliveryService(location, weight)

    price = delivery.calculate_price()

    output.delete("1.0", END)

    if price is not None:

        output.insert(END, f"Delivery Location: {location}\n")
        output.insert(END, f"Package Weight: {weight}kg\n")
        output.insert(END, f"Amount to Pay: N{price}")

    else:
        output.insert(END, "Invalid location entered.")


# GUI Window
root = Tk()
root.title("Delivery Charge System")
root.geometry("500x400")
root.config(bg="lightgreen")


# Heading
heading = Label(
    root,
    text="Delivery Charge Calculator",
    font=("Arial", 16, "bold"),
    bg="lightgreen"
)
heading.pack(pady=10)


# Location
location_label = Label(
    root,
    text="Enter Location (PAU or Epe):",
    font=("Arial", 12),
    bg="lightgreen"
)
location_label.pack()

location_entry = Entry(root, width=30)
location_entry.pack(pady=5)


# Weight
weight_label = Label(
    root,
    text="Enter Package Weight (kg):",
    font=("Arial", 12),
    bg="lightgreen"
)
weight_label.pack()

weight_entry = Entry(root, width=30)
weight_entry.pack(pady=5)


# Button
button = Button(
    root,
    text="Calculate",
    command=check_price,
    bg="darkgreen",
    fg="white"
)
button.pack(pady=15)


# Output Box
output = Text(root, width=50, height=8)
output.pack(pady=10)


# Run App
root.mainloop()
```
